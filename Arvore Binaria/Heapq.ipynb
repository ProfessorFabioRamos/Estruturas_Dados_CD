#Min-Heap:
## É uma árvore binária especial que é sempre preenchida da esquerda para a direita
O pai é sempre menor ou igual aos seus filhos. Isso significa que a raiz da árvore é sempre o menor elemento de todos. É a estrutura perfeita para filas de prioridade.

```python
import heapq
```

```python
# LIstas normais gerenciadas pelas funções da biblioteca heapq
meu_heap = []
# Inserindo elementos
heapq.heappush(meu_heap, 50)
heapq.heappush(meu_heap, 10)
heapq.heappush(meu_heap, 30)
heapq.heappush(meu_heap, 5)
```

```python
# a raiz é sempre o índice 0
print("A estrutura da lista Heap;", meu_heap)
```

```python
# Removendo elementos
#Acontece o "Sink Down" (reorganização da árvore para manter o menor como raiz)
print("Removido (menor prioridade):",heapq.heappop(meu_heap)) # 5
print("Removido (menor prioridade):",heapq.heappop(meu_heap)) # 10
print("Ficaram no Heap:",meu_heap) # 30,50
```
