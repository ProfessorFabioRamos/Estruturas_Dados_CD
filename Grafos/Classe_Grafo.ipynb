```python
class Grafo:
  def __init__(self, direcionado=False):
    # Direcionado = via de mão dupla
    self.direcionado = direcionado
    # Dicionario onde a chave é o node
    # e o valor é uma lista de tuplas (vizinho, peso)
    self.adjacencias = {}

  def adicionar_node(self, node):
    if node not in self.adjacencias:
      self.adjacencias[node] = []

  def adicionar_aresta(self, origem, destino, peso=1):
    self.adicionar_node(origem)
    self.adicionar_node(destino)
    # adiciona o destino na lista de origem
    self.adjacencias[origem].append((destino, peso))
    # Se não for direcionado faz o caminho inverso
    if not self.direcionado:
      self.adjacencias[destino].append((origem, peso))

  def obter_vizinhos(self,node):
    return self.adjacencias.get(node, [])
```

```python
rede_rotas = Grafo()
rede_rotas.adicionar_aresta("São Paulo", "Campinas", peso=90)
rede_rotas.adicionar_aresta("Campinas", "Ribeirão Preto", peso=220)
print(f"Vizinhos de Campinas: {rede_rotas.obter_vizinhos("Campinas")}")
```
