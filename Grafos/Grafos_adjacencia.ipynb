```python
# Matriz de Adjacência
# Matriz para 4 nós(nodes) (A,B,C,D)
# 1= conectado, 0= não conectado

matriz_adj = np.array([
    [0, 1, 1, 0],     # A Conecta com B e C
    [1, 0, 0, 1],     # B Conecta com A e D
    [1, 0, 0, 1],     # C Conecta com A e D
    [0, 1, 1, 0]      # D Conecta com B e C
])
```
