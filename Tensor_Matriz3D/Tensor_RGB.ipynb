```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

```python
# Criar um tensor 3D de zeros (preto) 16x16x3
imagem_rgb = np.zeros((16,16,3),dtype = np.uint8)
```

```python
# Fundo azul
imagem_rgb[:,:,2] = 255
# Quadrado vermelho no meio, primeiro removemos o azul
imagem_rgb[4:12,4:12,2] = 0
imagem_rgb[4:12,4:12,0] = 255
# Quadrado verde no meio, primeiro removemos o vermelho
imagem_rgb[7:9,7:9,0] = 0
imagem_rgb[7:9,7:9,1] = 255
```

```python
plt.imshow(imagem_rgb)
plt.axis('off')
plt.show()
```

```python
# Bandeira da França
bandeira = np.zeros((15,15,3),dtype = np.uint8)
# Parte azul
bandeira [:,0:5,2] = 255
# Parte branca
bandeira [:,5:10,0] = 255
bandeira [:,5:10,1] = 255
bandeira [:,5:10,2] = 255
# Parte vermelha
bandeira [:,10:15,0] = 255
#bandeira [:,10:15,1] = 255 # amarelo
```

```python
plt.imshow(bandeira)
#plt.axis('off')
plt.show()
```

```python
import cv2 as cv
# Lê a imagem com os 3 canais de cores em BGR
mush_matrix = cv.imread('mario_mush_16_16.png', cv.IMREAD_COLOR)
# Converte de BGR para RGB
mush_matrix = cv.cvtColor(mush_matrix, cv.COLOR_BGR2RGB)
print("Formato:",mush_matrix.shape)
```

```python
# Canal Red
print(mush_matrix[:,:,0])
```

```python
# Canal Green
print(mush_matrix[:,:,1])
```

```python
# Canal Blue
print(mush_matrix[:,:,2])
```
